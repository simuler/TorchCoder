# 参考题解：手写交叉熵损失

不调用 `F.cross_entropy`，从 logits 和类别索引手写数值稳定的交叉熵。

核心思路：先减去每行最大值，用 LogSumExp 得到稳定的 log-softmax，再 gather 目标类别并取负均值。


```python
# ✅ SOLUTION

import torch

def manual_cross_entropy(logits: torch.Tensor, targets: torch.Tensor) -> torch.Tensor:
    max_values = logits.max(dim=-1, keepdim=True).values
    logsumexp = torch.log(torch.exp(logits - max_values).sum(dim=-1, keepdim=True)) + max_values
    log_probs = logits - logsumexp
    return -log_probs.gather(-1, targets.unsqueeze(-1)).squeeze(-1).mean()
```
