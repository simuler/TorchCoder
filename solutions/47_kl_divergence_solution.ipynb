# 参考题解：KL 散度

从 logits 稳定计算离散分布 `KL(P || Q)`。

核心思路：对两组 logits 分别做 log-softmax，按 `sum p * (log p - log q)` 计算每个样本的 KL，并支持 none/mean/sum。


```python
# ✅ SOLUTION

import torch
import torch.nn.functional as F

def kl_divergence(p_logits: torch.Tensor, q_logits: torch.Tensor, reduction: str = "mean"):
    log_p = F.log_softmax(p_logits, dim=-1)
    log_q = F.log_softmax(q_logits, dim=-1)
    values = (log_p.exp() * (log_p - log_q)).sum(dim=-1)
    if reduction == "none": return values
    if reduction == "sum": return values.sum()
    if reduction == "mean": return values.mean()
    raise ValueError("reduction must be none, mean, or sum")
```
