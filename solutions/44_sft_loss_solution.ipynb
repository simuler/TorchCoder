# 参考题解：自回归 SFT 损失

实现支持 `ignore_index` 的语言模型监督微调损失。

核心思路：自回归模型用第 t 位 logits 预测第 t+1 位标签，因此 logits 去掉末位、labels 去掉首位，再展平计算交叉熵。


```python
# ✅ SOLUTION

import torch
import torch.nn.functional as F

def compute_sft_loss(logits: torch.Tensor, labels: torch.Tensor, ignore_index: int = -100):
    shift_logits = logits[..., :-1, :].contiguous()
    shift_labels = labels[..., 1:].contiguous()
    return F.cross_entropy(shift_logits.view(-1, shift_logits.size(-1)), shift_labels.view(-1), ignore_index=ignore_index)
```
